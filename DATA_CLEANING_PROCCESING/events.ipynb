```python
# τώρα θα φτιάξω μάπινγκ για τις χώρες και θα το περάσω σαν νέα στήλη 
```

```python
country_mapping = {}

for i, country in enumerate(data['country'].unique()):
    country_mapping[country] = i


data['country_encoded'] = data['country'].map(country_mapping)
```

```python
## country οκ πάμε για μήνες τώρα 
```

```python
month_mapping = {
    "January": 1,
    "February": 2,
    "March": 3,
    "April": 4,
    "May": 5,
    "June": 6,
    "July": 7,
    "August": 8,
    "September": 9,
    "October": 10,
    "November": 11,
    "December": 12
}


data['month_encoded'] =  data['event_month'].map(month_mapping)
```

```python
## πάμε τώρα πόλη 
```

```python
city_mapping = {}

for i, city in enumerate(data['city'].unique()):
    city_mapping[city] = i






data['city_encoded']  = data['city'].map(city_mapping)
```

```python
# ΤΏΡΑ ΤΑ REGION 
```

```python
region_mapping = {}

for i, reg in enumerate(data['region'].unique()):
    region_mapping[reg] = i

import json 




data['region_encoded']  = data['region'].map(region_mapping)
```
