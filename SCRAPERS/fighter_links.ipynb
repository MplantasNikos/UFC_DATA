```python
import requests
from bs4 import BeautifulSoup
import time
import pandas as pd

headers = {"User-Agent": "Mozilla/5.0"}

seen_fighters = set()
fighter_links_all = []

events_url = "http://ufcstats.com/statistics/events/completed?page=all"

events_html = requests.get(events_url, headers=headers).text
events_soup = BeautifulSoup(events_html, "html.parser")

event_links = [
    a["href"]
    for a in events_soup.select("a.b-link.b-link_style_black")
]

total_events = len(event_links)

print(f"\nTOTAL EVENTS: {total_events}\n")

for i, event_link in enumerate(event_links, start=1):

    print(f"\n EVENT {i}/{total_events}")
    print(event_link)

    time.sleep(0.5)

    event_html = requests.get(event_link, headers=headers).text
    event_soup = BeautifulSoup(event_html, "html.parser")

    fight_links = [
        a["href"]
        for a in event_soup.select("a.b-flag")
        if "fight-details" in a["href"]
    ]

    print(f"   FIGHTS FOUND: {len(fight_links)}")

    for j, fight_link in enumerate(fight_links, start=1):

        print(f"   ➜ Fight {j}/{len(fight_links)}")

        time.sleep(0.3)

        fight_html = requests.get(fight_link, headers=headers).text
        fight_soup = BeautifulSoup(fight_html, "html.parser")

        fighters = fight_soup.select("a.b-link.b-link_style_black")

        new_in_this_fight = 0

        for f in fighters:

            link = f.get("href")

            if not link:
                continue

            if "fighter-details" not in link:
                continue

            if link in seen_fighters:
                continue

            seen_fighters.add(link)
            fighter_links_all.append(link)

            new_in_this_fight += 1

        print(f"      NEW FIGHTERS THIS FIGHT: {new_in_this_fight}")
        print(f"      TOTAL UNIQUE FIGHTERS: {len(seen_fighters)}")

fighters_df = pd.DataFrame({
    "fighter_url": fighter_links_all
})

fighters_df.to_csv("ufc_fighter_links.csv", index=False)

print("DONE")
print("TOTAL UNIQUE FIGHTERS:", len(fighter_links_all))
print("CSV SAVED: ufc_fighter_links.csv")
```
