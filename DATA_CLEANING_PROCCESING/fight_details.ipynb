```python
## βλέπω ότι υπάρχει ποσοστο sig_str άρα θα μετατρέψω αυτό το ποσοστό σε δεκαδικο και θα κρατήσω απο όλα τα άλλα μόνο τα συνολικά χτυήματα που βρήκαν στόχο
```

```python
def get_first(column):
    xs = []
    for i  in  data[column]:
        x,y = i.split('of')
        x.strip()
        xs.append(x)
    name = column + '_total'
    data[name] = xs


cols = [
    'fighter1_sig_str',
    'fighter2_sig_str',

    'fighter1_head',
    'fighter2_head',

    'fighter1_body',
    'fighter2_body',

    'fighter1_leg',
    'fighter2_leg',

    'fighter1_distance',
    'fighter2_distance',

    'fighter1_clinch',
    'fighter2_clinch',

    'fighter1_ground',
    'fighter2_ground',

]
for col in cols:
    get_first(col)

    
```

```python
data.drop(columns = cols,inplace=True)
data.drop(columns = ['fighter1','fighter2','url'],inplace=True)
```

```python
data['fighter1_percent_sign'] = (
    pd.to_numeric(
        data['fighter1_sig_pct'].str.replace('%', ''),
        errors='coerce'
    ) / 100
)

data['fighter2_percent_sign'] = (pd.to_numeric(data['fighter2_sig_pct'].str.replace('%',''),errors='coerce')/100)
```
